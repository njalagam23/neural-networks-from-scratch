```python
import numpy as np

rng = np.random.default_rng(42)

# Generate 400 synthetic customers
n = 400

usage = rng.uniform(0, 40, n)
tenure = rng.integers(1, 61, n)
tickets = rng.poisson(1.5, n)

X = np.column_stack([usage, tenure, tickets])

# Simulated churn outcomes
logits = 0.9 - 0.14 * usage - 0.035 * tenure + 0.9 * tickets
true_probs = 1 / (1 + np.exp(-logits))

y = (rng.random(n) < true_probs).astype(float).reshape(-1, 1)

# Train/test split
idx = rng.permutation(n)
train_idx, test_idx = idx[:320], idx[320:]

X_train, X_test = X[train_idx], X[test_idx]
y_train, y_test = y[train_idx], y[test_idx]

# Standardize using training statistics only
mean = X_train.mean(axis=0)
std = X_train.std(axis=0)

X_train = (X_train - mean) / std
X_test = (X_test - mean) / std

print("Training:", X_train.shape)
print("Testing:", X_test.shape)
```

```python
import torch
import torch.nn as nn
import numpy as np
import matplotlib.pyplot as plt

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

torch.manual_seed(42)

X_train_t = torch.tensor(X_train, dtype=torch.float32)
X_test_t = torch.tensor(X_test, dtype=torch.float32)

y_train_t = torch.tensor(y_train, dtype=torch.float32)
y_test_t = torch.tensor(y_test, dtype=torch.float32)

print(X_train_t.shape)
print(y_train_t.shape)
```

```python
class ChurnModel(nn.Module):

    def __init__(self):
        super().__init__()

        self.hidden = nn.Linear(3, 8)
        self.output = nn.Linear(8, 1)

    def forward(self, x):

        x = self.hidden(x)
        x = torch.relu(x)

        x = self.output(x)

        return x


model = ChurnModel()

print(model)
```

```python
def train_model(model, X, y, epochs=600, lr=0.2):

    criterion = nn.BCEWithLogitsLoss()

    optimizer = torch.optim.SGD(
        model.parameters(),
        lr=lr
    )

    history = []

    model.train()

    for epoch in range(epochs):

        logits = model(X)

        loss = criterion(logits, y)

        optimizer.zero_grad()

        loss.backward()

        optimizer.step()

        history.append(loss.item())

        if (epoch + 1) % 100 == 0:
            print(
                f"Epoch {epoch+1}: "
                f"Loss = {loss.item():.4f}"
            )

    return history
```

```python
loss_history = train_model(
    model,
    X_train_t,
    y_train_t
)
```

```python
model.eval()

with torch.no_grad():

    test_logits = model(X_test_t)

    probabilities = torch.sigmoid(test_logits)

    predictions = (
        probabilities >= 0.5
    ).int().numpy().flatten()

actual = y_test_t.int().numpy().flatten()

accuracy = accuracy_score(actual, predictions)
precision = precision_score(actual, predictions, zero_division=0)
recall = recall_score(actual, predictions, zero_division=0)
f1 = f1_score(actual, predictions, zero_division=0)

print("Accuracy:", round(accuracy, 4))
print("Precision:", round(precision, 4))
print("Recall:", round(recall, 4))
print("F1 Score:", round(f1, 4))

print("\nConfusion Matrix:")
print(confusion_matrix(actual, predictions))

print("\nClassification Report:")
print(classification_report(actual, predictions, zero_division=0))
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
import numpy as np

# Recreate the original synthetic dataset
rng = np.random.default_rng(42)
n = 400

usage = rng.uniform(0, 40, n)
tenure = rng.integers(1, 61, n)
tickets = rng.poisson(1.5, n)

X = np.column_stack([usage, tenure, tickets])

logits = (
    0.9
    - 0.14 * usage
    - 0.035 * tenure
    + 0.9 * tickets
)

true_probs = 1 / (1 + np.exp(-logits))
y = (rng.random(n) < true_probs).astype(int)

# First: reserve 20% for testing
X_train_val, X_test, y_train_val, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Second: take 25% of the remaining 80%
# That gives 60% train, 20% validation, 20% test
X_train, X_val, y_train, y_val = train_test_split(
    X_train_val, y_train_val,
    test_size=0.25,
    random_state=42,
    stratify=y_train_val
)

# Fit scaling ONLY on training data
scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_val = scaler.transform(X_val)
X_test = scaler.transform(X_test)

print("Training:", X_train.shape)
print("Validation:", X_val.shape)
print("Testing:", X_test.shape)
```

```python
import torch
import torch.nn as nn
import matplotlib.pyplot as plt

torch.manual_seed(42)

X_train_t = torch.tensor(X_train, dtype=torch.float32)
X_val_t = torch.tensor(X_val, dtype=torch.float32)

y_train_t = torch.tensor(y_train, dtype=torch.float32).reshape(-1, 1)
y_val_t = torch.tensor(y_val, dtype=torch.float32).reshape(-1, 1)

print(X_train_t.shape)
print(y_train_t.shape)
```

```python
class ChurnModel(nn.Module):

    def __init__(self):
        super().__init__()

        self.hidden = nn.Linear(3, 8)
        self.output = nn.Linear(8, 1)

    def forward(self, x):
        x = torch.relu(self.hidden(x))
        x = self.output(x)
        return x

model = ChurnModel()
```

```python
criterion = nn.BCEWithLogitsLoss()
optimizer = torch.optim.SGD(model.parameters(), lr=0.2)

train_losses = []
val_losses = []

for epoch in range(600):

    # Training
    model.train()

    train_logits = model(X_train_t)
    train_loss = criterion(train_logits, y_train_t)

    optimizer.zero_grad()
    train_loss.backward()
    optimizer.step()

    # Validation
    model.eval()

    with torch.no_grad():
        val_logits = model(X_val_t)
        val_loss = criterion(val_logits, y_val_t)

    train_losses.append(train_loss.item())
    val_losses.append(val_loss.item())

    if (epoch + 1) % 100 == 0:
        print(
            f"Epoch {epoch+1} | "
            f"Train Loss: {train_loss.item():.4f} | "
            f"Validation Loss: {val_loss.item():.4f}"
        )
```

```python
import torch
import torch.nn as nn

class RegularizedChurnModel(nn.Module):
    def __init__(self, hidden_size=8, dropout_rate=0.2):
        super().__init__()

        self.hidden = nn.Linear(3, hidden_size)
        self.dropout = nn.Dropout(dropout_rate)
        self.output = nn.Linear(hidden_size, 1)

    def forward(self, x):
        x = torch.relu(self.hidden(x))
        x = self.dropout(x)
        return self.output(x)

model = RegularizedChurnModel(
    hidden_size=8,
    dropout_rate=0.2
)

print(model)
```

```python
import torch
import torch.nn as nn
import copy

torch.manual_seed(42)

regularized_model = RegularizedChurnModel(
    hidden_size=8,
    dropout_rate=0.2
)

criterion = nn.BCEWithLogitsLoss()

optimizer = torch.optim.SGD(
    regularized_model.parameters(),
    lr=0.2,
    weight_decay=0.001
)

best_val_loss = float("inf")
best_weights = None
best_epoch = 0

patience = 30
counter = 0

train_losses = []
val_losses = []

for epoch in range(600):

    # Training
    regularized_model.train()

    optimizer.zero_grad()

    logits = regularized_model(X_train_t)

    train_loss = criterion(logits, y_train_t)

    train_loss.backward()

    optimizer.step()

    # Validation
    regularized_model.eval()

    with torch.no_grad():
        val_logits = regularized_model(X_val_t)
        val_loss = criterion(val_logits, y_val_t)

    train_losses.append(train_loss.item())
    val_losses.append(val_loss.item())

    # Early stopping
    if val_loss.item() < best_val_loss - 1e-6:
        best_val_loss = val_loss.item()
        best_weights = copy.deepcopy(
            regularized_model.state_dict()
        )
        best_epoch = epoch + 1
        counter = 0
    else:
        counter += 1

    if counter >= patience:
        print("Early stopping at epoch:", epoch + 1)
        break

# Restore the best model
regularized_model.load_state_dict(best_weights)
regularized_model.eval()

print("Best Epoch:", best_epoch)
print("Best Validation Loss:", round(best_val_loss, 4))
print("Total Epochs Trained:", epoch + 1)
```

```python
def train_experiment(dropout_rate, weight_decay):

    torch.manual_seed(42)

    model = RegularizedChurnModel(
        hidden_size=8,
        dropout_rate=dropout_rate
    )

    criterion = nn.BCEWithLogitsLoss()

    optimizer = torch.optim.SGD(
        model.parameters(),
        lr=0.2,
        weight_decay=weight_decay
    )

    best_loss = float("inf")
    best_epoch = 0
    best_weights = None
    counter = 0

    for epoch in range(600):

        model.train()
        optimizer.zero_grad()

        logits = model(X_train_t)
        loss = criterion(logits, y_train_t)

        loss.backward()
        optimizer.step()

        model.eval()

        with torch.no_grad():
            val_loss = criterion(
                model(X_val_t),
                y_val_t
            ).item()

        if val_loss < best_loss - 1e-6:
            best_loss = val_loss
            best_epoch = epoch + 1
            best_weights = copy.deepcopy(model.state_dict())
            counter = 0
        else:
            counter += 1

        if counter >= 30:
            break

    model.load_state_dict(best_weights)
    model.eval()

    return model, best_loss, best_epoch
```

```python
experiments = [
    (0.0, 0.0),
    (0.1, 0.0),
    (0.2, 0.0),
    (0.0, 0.001)
]

experiment_results = []

for dropout, weight_decay in experiments:

    trained_model, val_loss, best_epoch = train_experiment(
        dropout,
        weight_decay
    )

    experiment_results.append({
        "dropout": dropout,
        "weight_decay": weight_decay,
        "val_loss": round(val_loss, 4),
        "best_epoch": best_epoch
    })

for result in experiment_results:
    print(result)
```

```python
import numpy as np
import torch

def predict_churn(usage, tenure, tickets):

    # Step 1: Prepare input features
    customer = np.array(
        [[usage, tenure, tickets]],
        dtype=np.float32
    )

    # Step 2: Apply training-time scaling
    customer_scaled = scaler.transform(customer)

    # Step 3: Convert to PyTorch tensor
    customer_tensor = torch.tensor(
        customer_scaled,
        dtype=torch.float32
    )

    # Step 4: Generate prediction
    loaded_model.eval()

    with torch.no_grad():
        logit = loaded_model(customer_tensor)
        probability = torch.sigmoid(logit).item()

    # Step 5: Apply validation-selected threshold
    threshold = 0.3

    prediction = (
        "High Churn Risk"
        if probability >= threshold
        else "Low Churn Risk"
    )

    return {
        "churn_probability": round(probability, 4),
        "risk_level": prediction
    }
```
