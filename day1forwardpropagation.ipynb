```python
import numpy as np
X = np.array ([[2,3]])
W = np.array([[0.5, -0.2], [0.3, 0.8]])
b = np.array([0.1, 0.2])
Z = X @ W + b
A = np.maximum(0, Z)
print("Weighted sum:", Z)
print("After ReLU:", A)
```

```python

import numpy as np

X = np.array([[2, 3]])

# First layer: 2 inputs -> 3 neurons
W1 = np.array([
    [0.5, -0.2, 0.4],
    [0.3, 0.8, -0.1]
])
b1 = np.array([0.1, 0.2, 0.3])

# Hidden layer with ReLU
hidden = np.maximum(0, X @ W1 + b1)

# Second layer: 3 neurons -> 1 output
W2 = np.array([[0.2], [0.5], [-0.3]])
b2 = np.array([0.1])

# Output with sigmoid
z = hidden @ W2 + b2
output = 1 / (1 + np.exp(-z))

print("Hidden layer:", hidden)
print("Final output:", output)
```

```python
import numpy as np

X = np.array([
    [12, 70, 8],
    [2, 95, 2],
    [36, 45, 15],
    [6, 80, 5],
    [24, 60, 12]
], dtype=float)

# Normalize the inputs
X = X / np.array([36, 100, 15])

print("Input shape:", X.shape)
print(X)
```

```python
rng = np.random.default_rng(42)

# Three inputs -> four hidden neurons
W1 = rng.normal(size=(3, 4))
b1 = np.zeros(4)

# Four hidden neurons -> one output
W2 = rng.normal(size=(4, 1))
b2 = np.zeros(1)

def relu(x):
    return np.maximum(0, x)

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Forward propagation
hidden = relu(X @ W1 + b1)
output = sigmoid(hidden @ W2 + b2)

print("Hidden layer shape:", hidden.shape)
print("Output shape:", output.shape)
print("Predictions:", output)
```
